Praktikum D1 — Memulai Klasifikasi Gambar dengan Dataset Sederhana (MNIST)

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist

# 1. Load data MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 2. Tampilkan contoh 9 gambar pertama
plt.figure(figsize=(5, 5))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

Praktikum D2 — Klasifikasi Gambar dengan Model Machine Learning Tradisional (SVM)

```python
from sklearn import svm
from sklearn.metrics import accuracy_score

# 1. Flatten gambar 28x28 menjadi vektor 784 dan normalisasi nilai pixel ke [0, 1]
x_train_flat = x_train.reshape(len(x_train), -1) / 255.0
x_test_flat = x_test.reshape(len(x_test), -1) / 255.0

# 2. Inisialisasi & latih model SVM (menggunakan 5000 sampel pertama agar proses cepat)
clf = svm.SVC(kernel='linear', gamma='scale')
clf.fit(x_train_flat[:5000], y_train[:5000])

# 3. Prediksi dan evaluasi akurasi
y_pred_svm = clf.predict(x_test_flat)
print("Akurasi SVM:", accuracy_score(y_test, y_pred_svm))
```

Praktikum D3 — Membangun CNN Sederhana (MNIST)

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Reshape data menjadi (samples, height, width, channels) dan normalisasi
x_train_cnn = x_train.reshape(-1, 28, 28, 1) / 255.0
x_test_cnn = x_test.reshape(-1, 28, 28, 1) / 255.0

# 2. Bangun arsitektur CNN Sederhana
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 3. Kompilasi model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 4. Latih model
history = model.fit(x_train_cnn, y_train, epochs=5, validation_split=0.1)

# 5. Visualisasi Grafik Akurasi dan Loss
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('CNN Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

Praktikum D4 — Eksperimen dengan Dataset Lebih Kompleks (CIFAR-10)

```python
from tensorflow.keras.datasets import cifar10
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Load dataset CIFAR-10
(x_train_cifar, y_train_cifar), (x_test_cifar, y_test_cifar) = cifar10.load_data()
x_train_cifar, x_test_cifar = x_train_cifar / 255.0, x_test_cifar / 255.0

# 2. Bangun arsitektur CNN yang lebih dalam
model_cifar = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 3. Kompilasi & Pelatihan
model_cifar.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_cifar = model_cifar.fit(x_train_cifar, y_train_cifar, epochs=10, validation_split=0.1)

# 4. Plot Grafik History CIFAR-10
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history_cifar.history['accuracy'], label='Training Accuracy')
plt.plot(history_cifar.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN CIFAR-10 Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_cifar.history['loss'], label='Training Loss')
plt.plot(history_cifar.history['val_loss'], label='Validation Loss')
plt.title('CNN CIFAR-10 Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

Praktikum D5 — Transfer Learning dengan Model Pra-Latih (VGG16)

```python
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Load model VGG16 tanpa top layer
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(32, 32, 3))
base_model.trainable = False  # Bekukan bobot base_model

# 2. Tambahkan layer classifier baru
model_tl = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 3. Kompilasi & Latih
model_tl.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_tl = model_tl.fit(x_train_cifar, y_train_cifar, epochs=5, validation_split=0.1)

# 4. Plot Hasil Transfer Learning
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history_tl.history['accuracy'], label='Training Accuracy')
plt.plot(history_tl.history['val_accuracy'], label='Validation Accuracy')
plt.title('Transfer Learning Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_tl.history['loss'], label='Training Loss')
plt.plot(history_tl.history['val_loss'], label='Validation Loss')
plt.title('Transfer Learning Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

Praktikum D6 — Evaluasi dengan Confusion Matrix dan Metrik Lain

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Prediksi data uji menggunakan model CNN MNIST (D3)
y_pred_cnn = model.predict(x_test_cnn).argmax(axis=1)

# 2. Cetak Classification Report (Precision, Recall, F1-Score)
print("Classification Report:")
print(classification_report(y_test, y_pred_cnn))

# 3. Tampilkan Confusion Matrix
cm = confusion_matrix(y_test, y_pred_cnn)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

Penugasan Praktikum — Pengujian Gambar Tulisan Tangan Sendiri

```python
# ===== 1) Upload File Foto =====
from google.colab import files
uploaded = files.upload()

# ===== 2) Fungsi Preprocessing Gambar ke Format MNIST =====
import numpy as np
from PIL import Image, ImageOps

def preprocess_to_mnist_28x28(img_pil):
    # Grayscale + Autocontrast
    img = img_pil.convert('L')
    img = ImageOps.autocontrast(img)
    arr = np.array(img).astype(np.uint8)

    # Invert warna jika latar belakang terang (kertas putih)
    if arr.mean() > 127:
        img = ImageOps.invert(img)
        arr = np.array(img)

    # Binarisasi ringan untuk menemukan bounding box digit
    thr = np.mean(arr) * 0.8
    mask = arr > thr
    if mask.any():
        ys, xs = np.where(mask)
        y0, y1 = ys.min(), ys.max()
        x0, x1 = xs.min(), xs.max()
        img = img.crop((x0, y0, x1 + 1, y1 + 1))

    # Resize ke 20x20 dengan mempertahankan rasio aspek
    img.thumbnail((20, 20), Image.Resampling.LANCZOS)

    # Pad ke 28x28 dan posisikan digit di tengah (center)
    canvas = Image.new('L', (28, 28), color=0)
    canvas.paste(img, ((28 - img.size[0]) // 2, (28 - img.size[1]) // 2))

    # Normalisasi ke [0, 1] dan tambah dimensi channel
    arr_out = np.array(canvas).astype('float32') / 255.0
    arr_out = arr_out[..., None]

    return canvas, arr_out

# ===== 3) Prediksi Menggunakan Model CNN (D3) =====
import matplotlib.pyplot as plt

results = []
for fname in uploaded.keys():
    img_pil = Image.open(fname)
    disp, x = preprocess_to_mnist_28x28(img_pil)

    # Expand dims menjadi (1, 28, 28, 1)
    x_batch = np.expand_dims(x, axis=0)

    probs = model.predict(x_batch, verbose=0)[0]
    pred = int(np.argmax(probs))
    conf = float(np.max(probs))
    results.append((fname, pred, conf))

    # Tampilkan Hasil Visualisasi
    plt.figure(figsize=(3, 3))
    plt.imshow(disp, cmap='gray')
    plt.title(f"Prediksi: {pred} (p={conf:.2f})")
    plt.axis('off')
    plt.show()

# Ringkasan Prediksi
print("Rekap Prediksi (CNN):")
for r in results:
    print(f"- {r[0]} -> {r[1]} (p={r[2]:.3f})")
```
